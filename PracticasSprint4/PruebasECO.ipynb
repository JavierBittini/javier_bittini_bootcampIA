```python
import requests

# la direccion base de nuestra API de pruebas
ECO = "https://deck.tech-thebridge.com/api/eco"

r = requests.get(ECO + "/get")

print(r.status_code)   # 200
print(r.ok)            # True si el codigo es 2xx

print(r.headers.get("Content-Type"))

print(r.text)          # el cuerpo, como texto
print(r.json())        # el cuerpo, como datos de Python
```
